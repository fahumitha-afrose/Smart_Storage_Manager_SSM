```python
import pandas as pd
import numpy as np
from google.colab import files
```

```python
file_list = []
np.random.seed(42)
```

```python
for i in range(100):
    file_size = np.random.randint(1, 200)
    modified = np.random.randint(1, 365)
    accessed = modified + np.random.randint(0, 30)
    duplicates = np.random.randint(0, 4)

    if(duplicates >= 3) or (file_size > 150 and accessed > 300):
        action = "CLEANUP"
    elif (duplicates == 2 and file_size > 50) or (file_size > 100 and accessed > 180):
        action = "REVIEW"
    else:
        action = "KEEP"

    file_list.append({"file_size_mb": file_size,
        "days_since_modified": modified,
        "days_since_accessed": accessed,
        "duplicate_count": duplicates,
        "action_label": action
    })
```

```python
df_synthetic = pd.DataFrame(file_list)
df_synthetic.to_csv("training_data.csv", index=False)
```

```python
print(df_synthetic.head(10))
```

```python
print("\nClass Distribution (%):")
total = len(df_synthetic)
for label, count in df_synthetic['action_label'].value_counts().items():
    print(f"{label}: {count} ({count/total*100:.1f}%)")
```
